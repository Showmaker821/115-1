```python
def calculate_discount(total):
    return total * 0.9

def get_total():
    p = int(input("請輸入單價："))
    q = int(input("請輸入數量："))
    total = p * q
    discounted = calculate_discount(total)
    print(f"原價金額：{int(total)}元")
    print(f"9折金額：{int(discounted)}元")

get_total()
```
