```python
def calculate_price(price, quantity):
    subtotal = price * quantity
    tax = subtotal * 0.05
    total = subtotal + tax
    return subtotal, tax, total

unit_price = int(input("請輸入單價："))
amount = int(input("請輸入數量："))

sub, tax_amount, total_amount = calculate_price(unit_price, amount)

print(f"小計: {sub}")
print(f"稅金: {tax_amount}")
print(f"總金額: {total_amount}")
```
